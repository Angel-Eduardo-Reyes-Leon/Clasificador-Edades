## Ejemplos de cosas útiles para hacer aquí:
- Visualizar imágenes random de cada bucket de edad
- Histograma de edades por dataset
- Distribución de etnia por bucket
- Inspeccionar features (LBP, HOG) de imágenes específicas
- Probar modelos con hiperparámetros distintos
- Visualizar curvas de aprendizaje
- Casos donde el modelo falla (predicción ≠ real)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from pathlib import Path

# Cargar datos
manifest = pd.read_csv('../data/final/manifest.csv')
features = pd.read_csv('../data/final/features.csv')
df = manifest.merge(features, on='filename')

print(f'Total: {len(df):,} imágenes')
print(f'Splits: {df["split"].value_counts().to_dict()}')
```

```python
# Distribución de edades por dataset
fig, ax = plt.subplots(figsize=(10, 5))
for source in manifest['source'].unique():
    sub = manifest[manifest['source'] == source]
    ax.hist(sub['age'], bins=30, alpha=0.5, label=source)
ax.set_xlabel('Edad')
ax.set_ylabel('Conteo')
ax.legend()
ax.set_title('Distribución de edad por dataset')
plt.show()
```
